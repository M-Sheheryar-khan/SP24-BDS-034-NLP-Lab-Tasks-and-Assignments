## Task 1: Regex Fundamentals (Character Classes & Quantifiers)

```python
import re

text = """
Ali visited Islamabad in 2025 with 3 friends.
They bought 12 books and spent 4500 rupees at the Market.
Python is fun, and Data Science is interesting!
"""

print(text)
```

### a) Extract all digits from the text

```python
digits = re.findall(r'\d', text)
print("Digits:", digits)
```

### b) Extract all words that start with a capital letter

```python
capital_words = re.findall(r'\b[A-Z][a-zA-Z]*\b', text)
print("Words starting with a capital letter:", capital_words)
```

### c) Extract all sequences of exactly 4 digits

```python
four_digits = re.findall(r'\b\d{4}\b', text)
print("Four-digit sequences:", four_digits)
```

### d) Extract all words that are between 3 to 6 characters long

```python
words_3_to_6 = re.findall(r'\b[A-Za-z]{3,6}\b', text)
print("Words between 3 and 6 characters:", words_3_to_6)
```

## Task 2: Pattern-Based Extraction (Real World Data)

```python
import re

text2 = """
Ali placed an order on 15/09/2026 and provided his email
ali.khan@gmail.com and phone number 0301-1234567.
Sara placed another order on 21/09/2026. Her email is
sara123@yahoo.com and her phone number is 0333-7654321.
The first order received a 45% discount, while the second
received a 12.5% discount.
"""
```

### a) Extract all email addresses

```python
emails = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', text2)
print("Email addresses:", emails)
```

### b) Extract all phone numbers in 03XX-XXXXXXX format

```python
phones = re.findall(r'\b03\d{2}-\d{7}\b', text2)
print("Phone numbers:", phones)
```

### c) Extract all dates in DD/MM/YYYY format

```python
dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', text2)
print("Dates:", dates)
```

### d) Extract all percentage values

```python
percentages = re.findall(r'\b\d+(?:\.\d+)?%', text2)
print("Percentages:", percentages)
```

## Task 3: Word Tokenization (NLTK vs spaCy)

```python
import nltk
import spacy

nltk.download('punkt')
nltk.download('punkt_tab')

paragraph = """
It's a beautiful day, but I don't want to go outside.
I'm working on my NLP assignment today.
Python isn't always easy, but it's very useful.
I don't mind learning new things when they're interesting!
"""
```

### a) Tokenize using NLTK

```python
from nltk.tokenize import word_tokenize

nltk_tokens = word_tokenize(paragraph)

print("NLTK Tokens:")
print(nltk_tokens)
```

### b) Tokenize using a blank spaCy pipeline

```python
nlp = spacy.blank("en")
doc = nlp(paragraph)

spacy_tokens = [token.text for token in doc]

print("spaCy Tokens:")
print(spacy_tokens)
```

### c) Comparison:
#### **1.** NLTK and spaCy handle contractions differently. For example,
####    NLTK may split "don't" into "do" and "n't", while spaCy uses
####    its own tokenization rules for contractions.

#### **2.** spaCy preserves whitespace/newline tokens in some situations,
####    while NLTK's word_tokenize() normally does not return whitespace
####    as separate tokens.

### d) Count total tokens produced by each method

```python
print("Total NLTK tokens:", len(nltk_tokens))
print("Total spaCy tokens:", len(spacy_tokens))
```

## Task 4: Token Attributes (Categorizing Tokens)

### (a) Create spaCy Doc

```python
import spacy

nlp = spacy.blank("en")
sentence = "Ali paid $50 for 3 books and sent the receipt to ali@gmail.com."
doc = nlp(sentence)

print(doc)
```

### b) Print is_alpha value for each token


```python
for token in doc:
    print(token.text, "-> is_alpha:", token.is_alpha)
```

### c) Print like_num and like_email values

```python
for token in doc:
    print(
        token.text,
        "-> like_num:", token.like_num,
        "| like_email:", token.like_email
    )
```

### d) List tokens where is_currency is True

```python
currency_tokens = [token.text for token in doc if token.is_currency]
print("Currency tokens:", currency_tokens)
```

## Task 5: Sentence Segmentation (Handling Tricky Cases)

### (a) Load trained pipeline

```python
import spacy

nlp = spacy.load("en_core_web_sm")

paragraph = """
Dr. Ahmed works at a hospital in Islamabad. He treats many patients every day.
Mr. Ali visited him yesterday because he wasn't feeling well.
The doctor suggested some changes, e.g. getting more sleep and drinking more water.
"""

doc = nlp(paragraph)
```

### b) Split paragraph into individual sentences

```python
sentences = list(doc.sents)
for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence.text.strip()}")
```

### c) Print total number of sentences

```python
print("Total number of sentences:", len(sentences))
```

### d) Observation:
#### The abbreviations "Dr.", "Mr.", and "e.g." did not cause incorrect sentence breaks because spaCy recognized them correctly within their sentence context.
